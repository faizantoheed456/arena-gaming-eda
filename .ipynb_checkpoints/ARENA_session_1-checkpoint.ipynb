#### ARENA: Unified Gaming Communities Analytics Platform

**Project Vision:**  
ARENA is a data‑driven platform designed to ingest, explore, and deeply analyze data from global gaming communities. Our goal is to uncover patterns in community growth, engagement, revenue, tournament activity, and player demographics. This repository contains the complete EDA (Exploratory Data Analysis) pipeline, from raw CSV to actionable insights.

**Dataset:** `gaming_communities.csv`  
- **Rows:** 400 gaming communities  
- **Columns:** 28 (including member_count, active_players, genre, region, monthly_revenue_usd, win_rate_pct, sponsorship, tournament_hosted, etc.)  
- **Source:** Real‐world inspired data with intentional missing values, outliers, and inconsistencies for robust practice.
