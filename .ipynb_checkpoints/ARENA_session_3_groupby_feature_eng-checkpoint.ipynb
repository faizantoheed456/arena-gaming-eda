*This notebook completes the grouping, aggregation, and feature engineering phase. The enhanced dataset is ready for visualisation and statistical testing in Session 4.*

```python
#importing essential libraries
import pandas as pd
import numpy as np
```

```python
#loading the dataset
gaming_data = pd.read_csv('gaming_communities_data_cleaned.csv')
```

```python
genre = gaming_data.groupby(['genre'])
print('Genre with the highest average community size:')
genre['member_count'].mean().nlargest(1)
```

Genre with the highest average community size is Fighting (5596.437), followed by MOBA (3551.31), FPS (3242.57) and so on 

```python
community_tier = gaming_data.groupby(['community_tier'])
community_tier['tournaments_hosted'].sum()
```

Semi‑Pro communities host the most tournaments (525), followed by Professional (123). Amateur and Elite host very few.

```python
prim_platform = gaming_data.groupby(['primary_platform'])
print('Platform with the highest average active players:')
prim_platform['active_players'].median().nlargest(1)
```

Platform with the highest average of active players is Twitter/X (311.0), followed by Twitch (303.0), YouTube (254.5) and so on

```python
region = gaming_data.groupby(['region'])
region['avg_member_age'].agg(['min', 'max'])
```

The region with the youngest member average age is EU (europe) with minimum avg age of 14.0 and maximum avg age of 37.9

```python
language = gaming_data.groupby(['language'])
language['community_name'].count().sort_values(ascending=False)
```

The top three most common languages used by communities are English (141), Spanish (62) and Portuguese (32)

```python
gaming_data.groupby(['genre']).agg({
    'member_count': 'mean',
    'tournaments_hosted': 'sum',
    'top_player_rank': 'median'
})
```

```python
gaming_data.groupby(['country_of_origin'])['member_count'].sum().sort_values(ascending=False).iloc[:5]
```

```python
gaming_data.groupby(['community_tier'])['member_count'].std().sort_values(ascending=False)
```

```python
gaming_data.groupby(['genre'])['primary_game'].nunique().sort_values(ascending=False)
```

```python
gaming_data['discord_to_reddit_ratio'] = gaming_data['discord_server_size'] / gaming_data['reddit_community_size']

gaming_data['discord_to_reddit_ratio'] = gaming_data['discord_to_reddit_ratio'].replace([np.inf, -np.inf], 0).fillna(0)

gaming_data.groupby(['region'])['discord_to_reddit_ratio'].mean()
```

```python
gaming_data.groupby(['genre', 'community_tier']).size().reset_index(name='count')
```

```python
gaming_data.groupby(['region', 'primary_platform'])['active_players'].mean().reset_index(name='Average').sort_values(by='Average', ascending=False)
```

```python
gaming_data.groupby(['year_founded', 'genre'])['member_count'].sum().reset_index(name='Total')
```

```python
gaming_data.groupby(['language', 'community_tier'])['top_player_rank'].max().reset_index(name='Worst_rank').nlargest(n=1, columns='Worst_rank')
```

```python
idx = gaming_data.groupby(['genre'])['member_count'].idxmax()
gaming_data.loc[idx, ['genre', 'community_name', 'member_count']]
```

- In the **FPS** genre, "Reaper Crew" is the largest community (68,819 members).  
- The **Strategy** genre has the smallest largest community (657 members), reflecting its niche audience.  
- Fighting and Battle Royale also have very large communities (>50k members).

```python
gaming_data['region_activity_rank'] = gaming_data.groupby(['region'])['active_players'].rank(ascending=False, method='dense')
top_per_region = gaming_data[gaming_data['region_activity_rank'] == 1][['region', 'community_name', 'active_players']]
top_per_region
```

- **Asia (AS)** has the most active community ("Pulse Nation" with 23,727 active players).  
- **Unknown** region also shows high activity ("Sunfire" with 14,589).  
- **South America (SA)** has the lowest top activity (4,811), suggesting smaller active player bases.

```python
filtered = gaming_data.groupby(['genre']).filter(lambda x: x['tournaments_hosted'].sum() > 50)
print(f"Removed Genres: {set(gaming_data['genre']) - set(filtered['genre'])}")
```

- Genres removed (total tournaments ≤ 50): **MMORPG, Strategy, Fighting**.  
- These genres host very few tournaments overall, so they may be less competitive or have different community structures.

```python
sorted_data = gaming_data.sort_values(by=['community_tier', 'member_count'], ascending=[True, False])

second_per_tier = sorted_data.groupby('community_tier').nth(1).reset_index()

second_per_tier[['community_tier', 'community_name', 'member_count']]
```

- In the **Amateur** tier, the second largest community is "Dusk Raiders" (1,879 members).  
- For **Elite**, it's "Pulse Nation" (68,819) – note that Elite has only 3 communities, so second largest is actually the second best.  
- **Semi-Pro** second largest is "War Hammer" (13,712 members).


```python
region_avg = gaming_data.groupby(['region'])['member_count'].transform('mean')
gaming_data['above_region_avg'] = gaming_data['member_count'] > region_avg
print(f"Percentage above region average: {gaming_data['above_region_avg'].mean()*100:.1f}%")
```

- Only **18.8%** of communities have member count above their region's average.  
- This means most communities are smaller than the regional average – the average is pulled up by a few very large communities (outliers).  
- Regions with high inequality (e.g., one giant community) will have a low percentage.

```python
gaming_data['engagement_ratio'] = gaming_data['active_players'] / gaming_data['member_count']
gaming_data.groupby(['genre'])['engagement_ratio'].mean()
```

- **Battle Royale** and **MMORPG** have the highest engagement ratios (~0.45), meaning nearly half of their members are active players.  
- **Strategy** has the lowest engagement (0.37), possibly due to smaller, more casual communities.

```python
gaming_data['tier_score'] = gaming_data['community_tier'].map({
    'Amateur': 1, 
    'Semi-Pro': 2, 
    'Professional': 3, 
    'Elite': 4
})

gaming_data.groupby(['region'])['tier_score'].mean()
```

- **Asia (AS)** has the highest average tier score (1.63), meaning its communities are closest to Semi‑Pro on average.  
- **South America (SA)** has the lowest (1.37), indicating more Amateur-level communities.

```python
bins = [0, 18, 25, 35, 100]
labels = ['Teen', 'Young Adult', 'Adult', 'Senior']

gaming_data['age_group'] = pd.cut(gaming_data['avg_member_age'], bins=bins, labels=labels)
gaming_data.groupby(['genre'])['age_group'].value_counts().unstack()
```

- **FPS** is the most popular genre across all age groups, especially among Young Adults (108 communities).  
- **MMORPG** and **MOBA** have more Adult and Senior members compared to other genres.

```python
gaming_data['community_age'] = 2026 - gaming_data['year_founded'] 

idx_oldest = gaming_data.groupby(['region'])['community_age'].idxmax()
idx_youngest = gaming_data.groupby(['region'])['community_age'].idxmin()

oldest = gaming_data.loc[idx_oldest, ['region', 'community_name', 'community_age']].assign(status='Oldest')
youngest = gaming_data.loc[idx_youngest, ['region', 'community_name', 'community_age']].assign(status='Youngest')

pd.concat([oldest, youngest]).sort_values(by=['region', 'status'])
```

- The **oldest** communities (founded 2008–2010) are in **AF**, **AS**, **EU**, and **Unknown** regions.  
- The **youngest** communities (founded 2024) exist in every region, showing continuous growth.  
- **EU** has the widest age range (2 to 18 years).

```python
gaming_data['total_social_presence'] = gaming_data['discord_server_size'] + gaming_data['reddit_community_size'] + (gaming_data['content_creators'] * 1000)
gaming_data.nlargest(5, 'total_social_presence')[['community_name', 'total_social_presence']]
```

- **Pulse Nation** has the highest total social presence (123,596), driven by a massive Discord server (95,596) and many content creators (28 × 1000).  
- **Reaper Crew** and **Sunfire** also have >99k scores, indicating they are highly active across platforms.

```python
gaming_data['has_coaching'] = gaming_data['coaching_staff'] > 0
gaming_data.groupby(['community_tier'])['has_coaching'].mean() * 100
```

- **100%** of Elite and Professional communities have coaching staff.  
- **66.4%** of Semi‑Pro communities have coaching, while **0%** of Amateur communities do – coaching is a strong differentiator of competitive tiers.

```python
gaming_data['has_multiple_platforms'] = (gaming_data['discord_server_size'] > 0) & (gaming_data['reddit_community_size'] > 0)
print(f"Percentage with multiple platforms: {gaming_data['has_multiple_platforms'].mean()*100:.1f}%")
```

- Only **51.5%** of communities maintain both a Discord server and a Reddit presence. Most focus on a single primary platform.

```python
gaming_data['log_member_count'] = np.log1p(gaming_data['member_count'])
```

```python
gaming_data['tournament_intensity'] = gaming_data['tournaments_hosted'] / (gaming_data['member_count'] / 1000)
print(f"Mean tournament intensity: {gaming_data['tournament_intensity'].mean():.2f}")
```

```python
gaming_data['top_player_tier'] = pd.qcut(gaming_data['top_player_rank'], q=3, labels=['High', 'Medium', 'Low'])
gaming_data.groupby(['genre'])['top_player_tier'].value_counts().unstack()
```

- **FPS** has the most communities in the "Low" rank (worst) tier (60), but also many in "Medium" and "High".  
- **Battle Royale** and **MOBA** are more balanced across tiers.  
- **Fighting** and **Strategy** have few communities overall, with most in "Medium".

```python
pivot = pd.pivot_table(
    gaming_data,
    index='genre',
    columns='community_tier',
    values='member_count',
    aggfunc='mean',
    fill_value=0
)
pivot
```

```python
pivot = pd.pivot_table(
    gaming_data,
    index='region',
    columns='primary_platform',
    values='community_name',
    aggfunc='count',
    fill_value=0
)
pivot
```

```python
crosstab = pd.crosstab(
    gaming_data['genre'],
    gaming_data['language'],
    normalize='index'
)
crosstab
```

```python
gaming_data = gaming_data.sort_values(by=['genre', 'year_founded'])
gaming_data['cumulative_member_by_genre'] = gaming_data.groupby(['genre'])['member_count'].cumsum()
```

```python
if 'index' in gaming_data.columns:
    gaming_data = gaming_data.drop(columns=['index'])

# 2. Sort the dataset back by your community_id so it starts from 1 and goes perfectly in order
gaming_data = gaming_data.sort_values(by='community_id').reset_index(drop=True)

# 3. Save it again cleanly without any extra index columns tracking the old mess
gaming_data.to_csv('gaming_communities_enhanced.csv', index=False)
print("Dataset completely restored! 'index' column removed, and ordered from 1 onwards.")
```
