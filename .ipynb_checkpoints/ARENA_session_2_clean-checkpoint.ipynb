### 2.4 Key Design Decisions

*This notebook documents the complete cleaning pipeline – all steps are reproducible and commented.*

```python
#importing essential libraries
import numpy as np
import pandas as pd
```

```python
#loading the dataset 
gaming_data = pd.read_csv('gaming_communities.csv')
```

```python
#makeing a copy to work on
gaming_copy = gaming_data.copy()
```

```python
#displaying first 5 rows
gaming_copy.head()
```

```python
#sum of columns with null values
gaming_copy.isna().sum().sort_values(ascending=False)
```

twitch_stream_viewers_avg has 302 nulls (75.5%), followed by monthly_revenue_usd (302), win_rate_pct (288), etc.

```python
# dropping the rows where 'active_players' are null
gaming_copy.dropna(subset=['active_players'], inplace=True)
print(f"Total no of rows after dropping the 'active_players' null values: {gaming_copy.shape[0]}")
```

All the rows are dropped where the 'active_players' are Null. The remaining rows after dropping them are 357

```python
#filling the avg_member_age with median
gaming_copy['avg_member_age'] = gaming_copy['avg_member_age'].fillna(gaming_copy['avg_member_age'].median()) 
gaming_copy['avg_member_age'].isna().sum()
```

I filled up the null values of 'avg_member_age' with median which reduces the chance of anomalies and outliers.

```python
# filling up secondary_game column with 'None' string
gaming_copy['secondary_game'] = gaming_copy['secondary_game'].fillna('None')
gaming_copy['secondary_game'].isna().sum()
```

I filled up the null values of 'secondary_game' with 'None' because it is not neccessary for communities to have secondary game

```python
#dropping columns with null values more than 60% 
null_percentage = (gaming_copy.isna().sum() / len(gaming_copy)) * 100
columns_to_drop = null_percentage[null_percentage > 60].index
gaming_copy.drop(columns=columns_to_drop, inplace=True)
print(f'Dropped columns {list(columns_to_drop)}')
```

The four columns with more than 60% null values has been removed successfully. These columns had too little data to be useful, so removing them simplifies the dataset without losing critical information.

```python
#forward fill the 'year_founded' column and type cast it to int64
gaming_copy['year_founded'] = gaming_copy['year_founded'].ffill().astype('Int64')
gaming_copy['year_founded'].dtype
```

Forward fill propagated the most recent known year to missing values. However, this method can spread incorrect future years (e.g., 3005) if they appear earlier.

```python
#droping the duplicate rows from dataset
before = gaming_copy.shape[0]
print(f'No of rows before removing duplicates: {before}')
duplicate_rows = gaming_copy.duplicated().sum()
print(f'Duplicate rows found: {duplicate_rows}')
gaming_copy = gaming_copy.drop_duplicates()
gaming_copy.reset_index(drop=True, inplace=True)
after = gaming_copy.shape[0]
print(f'No of rows after removing duplicates: {after}')
```

There are no duplicate rows in our dataset so the number of rows will remain same.

```python
#standardize the sponsorship, verified_account, and recruitment_open columns from 'Yes'/'No' strings to boolean True/False.
gaming_copy[['sponsorship', 'verified_account', 'recruitment_open']] = gaming_copy[['sponsorship', 'verified_account', 'recruitment_open']].map(bool)
gaming_copy[['sponsorship', 'verified_account', 'recruitment_open']].dtypes
```

I have standardize the 'sponsorship', 'verified_account', and 'recruitment_open' columns from 'Yes'/'No' strings to boolean True/False.

```python
#drop rows where 'active_players' and 'avg_member_age' is Null
gaming_rows = gaming_copy.dropna(subset=['active_players', 'avg_member_age']).shape[0]
print(f"Total number of rows after dropping 'active_players' and 'avg_member_age' Null values: {gaming_rows}")
```

There will be no effect on dataset beacuse we already filled up the Null values of 'avg_member_age' with median, and already drop the rows containing Null values of 'active_players'

```python
#filling 'top_player_rank' with bfill()
gaming_copy['top_player_rank'] = gaming_copy['top_player_rank'].bfill()
gaming_copy['top_player_rank'].head()
```

Backward fill used to fill missing ranks. This assumes that the next community's rank is a reasonable substitute – a simplification for practice.

```python
#converting dtype of column 'coaching_staff' from int64 to float32 and reviwing the impact on memory
mem_before = gaming_copy['coaching_staff'].memory_usage(index=False)

gaming_copy['coaching_staff'] = gaming_copy['coaching_staff'].astype('float32')

mem_after = gaming_copy['coaching_staff'].memory_usage(index=False)

print(f"Memory before: {mem_before} bytes")
print(f"Memory after: {mem_after} bytes")
print(f"Total memory saved: {mem_before - mem_after} bytes")
```

After converting memory of 'coaching_staff' column the memory usage reduces at 2x.

```python
zero_variance_mask = gaming_copy.nunique() == 1
cols_to_drop = zero_variance_mask[zero_variance_mask].index

if len(cols_to_drop) > 0:
    gaming_copy.drop(columns=cols_to_drop, inplace=True)
    print(f"Dropped zero-variance columns: {list(cols_to_drop)}")
else:
    print("No zero-variance columns found")
```

There are three columns with zero-variance columns. Analysis would be much harder if we keep these types of columns.

```python
#replacing 0 in tounaments_hosted with nan
gaming_copy['tournaments_hosted'] = gaming_copy['tournaments_hosted'].replace(0, np.nan)
print(f"Total null values in 'tounaments_hosted' columns: {gaming_copy['tournaments_hosted'].isna().sum()}")
```

```python
#calculate the 99th percentile threshold
quantile_99 = gaming_copy['member_count'].quantile(0.99)

#Cap the extreme values at the upper bound
gaming_copy['member_count'] = gaming_copy['member_count'].clip(upper=quantile_99)

print(f"99th percentile threshold: {quantile_99:.2f}")
print(f"Maximum value after clipping: {gaming_copy['member_count'].max():.2f}")
```

```python
#fixing impossible future years 
gaming_copy.loc[gaming_copy['year_founded'] > 2026, 'year_founded'] = np.nan
```

The imposible 'years_founded' values are now resolved.

```python
#removing special characters, symbols or corrupted text punctuation from 'community_name' column
gaming_copy['community_name'] = gaming_copy['community_name'].str.replace(r'[^\w\s]', '', regex=True) 
```

The special characters, symbols or corrupted text has been removed from dataset

```python
#validate member logic (active_players <= member_count)
violation_mask = gaming_copy['active_players'] > gaming_copy['member_count']
num_violations = violation_mask.sum()

print(f"Number of rows where active players exceed member count: {num_violations}.")

if num_violations > 0:
    gaming_copy = gaming_copy[~violation_mask].reset_index(drop=True)
    print("Successfully dropped logical violations and reset index!")
```

```python
#find duplicate community names
duplicate_names_df = gaming_copy[gaming_copy.duplicated(subset=['community_name'], keep=False)]

print(f"Total rows with duplicate community names: {len(duplicate_names_df)}")

duplicate_names_df[['community_id', 'community_name', 'primary_game', 'region']].sort_values(by='community_id').head(10)
```

```python
#strip leading and trailing whitespace from all string columns

string_cols = gaming_copy.select_dtypes(include=['object']).columns

for col in string_cols:
    gaming_copy[col] = gaming_copy[col].str.strip()

print(f"Successfully stripped whitespace from text columns:\n{list(string_cols)}")
```

```python
#flag incomplete records
gaming_copy['incomplete_record_flag'] = gaming_copy.isnull().sum(axis=1) > 5
print(f"Total highly incomplete records found: {gaming_copy['incomplete_record_flag'].sum()}")
```

```python
#filling up all missing values

# region — still object dtype, safe to fill directly
gaming_copy['region'] = gaming_copy['region'].fillna('Unknown')

# country_of_origin may already be category dtype if cell below ran first
# Convert to object before filling so 'Unknown' can be inserted safely
gaming_copy['country_of_origin'] = (
    gaming_copy['country_of_origin']
    .astype(object)
    .fillna('Unknown')
)

# Clean up any leftover rank nulls at the edges (bfill may leave last rows null)
gaming_copy['top_player_rank'] = gaming_copy['top_player_rank'].fillna(
    gaming_copy['top_player_rank'].median()
)

# Forward-fill the spots left behind by the future-year fix
gaming_copy['year_founded'] = gaming_copy['year_founded'].ffill().astype('Int64')

# Fill missing platform sizes with 0 (no presence = 0)
size_cols = ['reddit_community_size', 'discord_server_size', 'content_creators', 'social_media_platforms']
gaming_copy[size_cols] = gaming_copy[size_cols].fillna(0)

gaming_copy['tournaments_hosted'] = gaming_copy['tournaments_hosted'].fillna(0).astype('int64')

print('All remaining nulls filled successfully.')
print(f'Null count after fill: {gaming_copy.isnull().sum().sum()}')
```

## 2.6 Data Cleaning Insights & Architectural Decisions

Now that the dataset has a 0-null footprint, here is the justification for the engineering choices made during this session to ensure analytical integrity:

### 1. Imputation Strategy & Logical Safeguards
* **Zero-Tournament Rules:** Missing values in `tournaments_hosted` were filled with `0` and downcast to `int64`. This structurally ensures that communities with no competitive footprint don't accidentally skew average team metrics.
* **Platform Presence:** Missing sizes for `reddit_community_size` and `discord_server_size` were set to `0` under the business assumption that a missing value implies a non-existent community presence on that specific platform rather than unrecorded data.
* **Categorical Fallbacks:** Remaining missing structural identifiers in `region` and `country_of_origin` were assigned to an explicit `'Unknown'` category to prevent complete row loss while avoiding artificial bias.

### 2. Risk Mitigation & Order of Operations
* **The Forward-Fill (`.ffill()`) Trap:** Forward-filling `year_founded` too early carries the systemic risk of propagating impossible future anomalies (like `3005`). Isolating and replacing future years with `NaN` *prior* to final structural filling preserved chronological accuracy.
* **String Integrity:** Applying regex pattern matching `[^\w\s]` to `community_name` stripped out decorative emojis and competitive tags (e.g., `★`, `#`, `!`), guaranteeing clean string keys for upcoming aggregation joins.

### 3. Outlier Stabilization
* **99th Percentile Clipping:** Rather than outright deleting high-growth outlier communities via trimming, `member_count` was capped at its 99th percentile ($68,819.36$). This retains the records for macro-analysis while heavily insulating variance-sensitive metrics from extreme skew.

```python
gaming_copy['community_id'] = range(1, len(gaming_copy) + 1)
```

```python
int_cols = [
    'member_count', 'active_players', 'top_player_rank', 'social_media_platforms',
    'discord_server_size', 'reddit_community_size', 'content_creators'
]
cat_cols = ['genre', 'country_of_origin', 'language', 'community_tier', 'primary_platform']

# coaching_staff is float32 — keep as is, don't cast to int64
gaming_copy[int_cols] = gaming_copy[int_cols].round().astype('int64')

# Convert string columns to category dtype for memory efficiency
gaming_copy[cat_cols] = gaming_copy[cat_cols].astype('category')

print('Dtype conversions complete.')
print(gaming_copy[int_cols + cat_cols].dtypes)
```

```python
# Save the final dataset
gaming_copy.to_csv('gaming_communities_data_cleaned.csv', index=False)
print("File successfully saved as 'gaming_communities_data_cleaned.csv'. You are ready for analysis!")
```
